```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

CSV = "/app/results/analyze_data/resub/return_data.csv"
LAYOUTS = ["Counter Circuit", "Coordination Ring"]

df = pd.read_csv(CSV)

# One row per (user, layout, agent); average each user's return across agents.
user_avg = (
    df[df["layout"].isin(LAYOUTS)]
    .groupby(["layout", "user_id"])["total_return"]
    .mean()
    .reset_index()
)

bins = np.histogram_bin_edges(user_avg["total_return"], bins=12)

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6), sharex=True, sharey=True)
for ax, layout in zip(axes, LAYOUTS):
    vals = user_avg.loc[user_avg["layout"] == layout, "total_return"]
    ax.hist(vals, bins=bins, color="#4f7fd6", edgecolor="white", linewidth=1.2)
    ax.axvline(vals.mean(), color="#333333", linestyle="--", linewidth=1.5)
    ax.set_title(f"{layout}  (n={len(vals)}, mean={vals.mean():.1f})", fontsize=11)
    ax.set_xlabel("Mean return across agents")
    ax.grid(axis="y", alpha=0.25, linewidth=0.6)
    ax.set_axisbelow(True)
    for side in ("top", "right"):
        ax.spines[side].set_visible(False)

axes[0].set_ylabel("Number of users")
fig.suptitle("Distribution of per-user average return", fontsize=13)
fig.tight_layout()
plt.show()
```
